```
public class Product
{
    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
        
    }
    private string name;
    public string Name
    {
        get{return name;}
        set{name = value;}
    }
    private decimal price;
    public decimal Price
    {
        get{return price;}
        set{price = value;}
    }
    private int quantity;
    public int Quantity
    {
        get{return quantity;}
        set{quantity = value;}
    }
    protected string category;
    public string Category
    {
        get{return category;}
        set{category = value;}
    }
    internal string description;
    public string Description
    {
        get{return description;}
        set{description = value;}
    }
    public decimal GetTotalPrice() //Возвращает общую стоимость всех единиц товара
    {
        return Price*Quantity;
    }
    public void UpdateQuantity(int amount) //Обновляет количество товара на указанное значение
    {
        Quantity = amount;
    }
    public string GetProductInfo() //Возвращает строку с полным описанием товара
    {
        return $"Товар {Name} категории {Category} с ценой {Price} в количестве {Quantity}. Описание товара: {Description}";
    }
}
public class Seller
{
    public Seller(string name, string employeeId, decimal salary, string contactinfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        Contactinfo = contactinfo;
        
    }
    private string name;
    public string Name
    {
        get{return name;}
        set{name = value;}
    }
    public string employeeId;
    public string EmployeeId
    {
        get{return employeeId;}
        set{employeeId = value;}
    }
    protected decimal salary;
    public decimal Salary
    {
        get{return salary;}
        set{salary = value;}
    }
    internal string contactinfo;
    public string Contactinfo
    {
        get{return contactinfo;}
        set{contactinfo = value;}
    }
    
    public List<Product> listofproducts = new List<Product>();
    
    public void AddProduct(Product product) //Добавить товар в список товаров продавца
    {
        listofproducts.Add(product);
    }
    public void SellProduct(Product product, int quantity) //Уменьшить количество товара на складе
    {
        product.Quantity -= quantity;
    }
    public string GetSellerInfo() //Вернуть информацию о продавце
    {
        return $"Продавец {Name } с должностью {EmployeeId} и зарплатой {Salary}. Контактная информация: {Contactinfo}";
    }
}
public class Store
{
    public Store(string storename, string location, string storehours)
    {
        Storename = storename;
        Location = location;
        Storehours = storehours;
        
    }
    private string storename;
    public string Storename
    {
        get{return storename;}
        set{storename=value;}
    }
    private string location;
    public string Location
    {
        get{return location;}
        set{location=value;}
    }
    private string storehours;
    public string Storehours
    {
        get{return storehours;}
        set{storehours=value;}
    }
    public List<Seller> listofsellers = new List<Seller>();
    public void AddSeller(Seller seller) //Добавить продавца в магазин
    {
        listofsellers.Add(seller);
    }
    public void ListProducts() //Перечислить товары в магазине
    {
        foreach (var seller in listofsellers)
        {
            foreach (var product in seller.listofproducts)
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }
    public string GetStoreInfo() //Вернуть информацию о магазине
    {
        return $"Магазин {Storename} с часами работы {Storehours} по адресу {Location}";
        
    }
}
Product myProduct_1 = new Product("Телефон сяоми", 12000, 34, "Смартфон", "Новая модель");
Product myProduct_2 = new Product("Ноутбук асус", 40000, 50, "Ноутбук", "Новая модель");
Console.WriteLine(myProduct_1.GetProductInfo());
Console.WriteLine(myProduct_1.GetTotalPrice());
myProduct_1.UpdateQuantity(400);
Seller mySeller = new Seller("Давид", "Менеджер", 12332, "88005553535");
Store myStore = new Store("Семёрочка", "Мельникайте 44", "8-20");
mySeller.AddProduct(myProduct_1);
mySeller.AddProduct(myProduct_2);
mySeller.SellProduct(myProduct_1, 34);
Console.WriteLine(mySeller.GetSellerInfo());
myStore.AddSeller(mySeller);
Console.WriteLine(myStore.GetStoreInfo());
myStore.ListProducts();

```
