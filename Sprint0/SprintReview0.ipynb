```
using System.Drawing;
using System.Net.Security;
using System.Reflection.Metadata;

public class Plant
{
    
    private string name;
    private string type;
    private float height;
    
    public string Name
    {
        get {return name;}
        set {name = value;}
    }
    public string Type
    {
        get {return type;}
        set {type = value;}
    }
    public float Height
    {
        get {return height;}
        set {height = value;}
    }
    public virtual void Blooming()
    {
        Console.WriteLine($"Растение {Name} цветёт");
    }
    public void Growing(float height)
    {
        Height = Height + height;
    }
}
public class Flower : Plant
{
    private string color;
    private int numberOfPetals;
    
    public string Color
    {
        get {return color;}
        set {color = value;}
    }
    public int NumberOfPetals
    {
        get {return numberOfPetals;}
        set {numberOfPetals = value;}
    }
    public override void Blooming()
    {
        Console.WriteLine($"Цветок {Name} расцветает и радует глаз!");
    }
    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать");
    }
}
public class Tree : Plant
{
    private int age;
    private string typeOfLeaves;
    public int Age
    {
        get {return age;}
        set {age = value;}
    }
    public string TypeOfLeaves
    {
        get {return typeOfLeaves;}
        set {typeOfLeaves = value;}
    }
    public override void Blooming()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }
    public void Increase(int age)
    {
        Age = Age + age;
        Console.WriteLine($"Новый возраст дерева - {Age}");
    }
}    

Flower myFlower = new Flower();
myFlower.Name = "Роза";
myFlower.Type = "Rose";
myFlower.Height = 20;
myFlower.Color = "Red";
myFlower.NumberOfPetals = 10;
myFlower.Blooming();
myFlower.Growing(30);
Console.WriteLine(myFlower.Height);
Console.WriteLine("------------------------------");
myFlower.Wither();
Tree myTree = new Tree();
myTree.Name = "Сосна";
myTree.Type = "Хвойная";
myTree.Height = 400;
myTree.Age = 20;
myTree.TypeOfLeaves = "Хвоя";
myTree.Increase(1);
Console.WriteLine(myTree.Height);
myTree.Blooming();



```
