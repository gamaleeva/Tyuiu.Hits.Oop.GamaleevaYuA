Описание задачи: 
Создать базовый класс Employee в C#, который будет представлять информацию о 
сотрудниках компании. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу Employee: 
• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary). 
• Методы: 
o GetInfo(): метод для получения информации о сотруднике в виде 
строки. 
o Work(): метод для вывода сообщения о выполнении рабочих 
обязанностей. 
o TakeVacation(): метод для вывода сообщения о взятии отпуска. 
Требования к производным классам: 
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие 
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть 
переопределен для добавления информации о управлении командой. 
2. Разработчик (Developer): Должен содержать дополнительные атрибуты, 
такие как Специализация (Specialization). Метод TakeVacation() должен быть 
переопределен для добавления информации о необходимости согласования 
отпуска с проектными сроками. 
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle). 
Метод Work() должен быть переопределен для добавления информации о 
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```

using System;
using System.Collections.Generic;
public class Employee 
{
   

    private string name;
    private int age;
    private int salary;
    protected string Name
    {
        get {return name;}
        set {name = value;}
    }
    protected int Age
    {
        get {return age;}
        set {if (value > 14) age = value;
        else Console.WriteLine("Возраст сотрудника не может быть меньше 14 лет");}
    }
    protected int Salary
    {
        get {return salary;}
        set {salary = value;}
    }
    public Employee(string name, int age)
    {
        Name = name;
        Age = age;
        Salary = 27000;
    }
    
    
    public void GetInfo()
    {
        Console.WriteLine($"Имя сотрудника: {Name}. Возраст: {Age}. Зарплата: {Salary}");
    }
    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие обязанности");
    }
    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} взял отпуск !");
    }
}
public class Manager : Employee
{
    public Manager(string name, int age) : base(name, age)
    {
      SubordinatesCount = 0;
    }
    
    private int subordinatescount;
    public int SubordinatesCount
    {
        get {return subordinatescount;}
        set {subordinatescount = value;}
    }
    public override void Work() // Перегрузка методов
    {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек");
    }
    public void Work(int subordinates) // Перегрузка методов
    {
        SubordinatesCount += subordinates;
        Work();
    }
   
}
public class Developer : Employee
{
    public Developer(string name, int age) : base (name, age)
    {
        Qualification= "Junior";
        Specialization = "";
    }
    private string specialization;
    private string qualification;
    public string Specialization
    {
        get {return specialization;}
        set {specialization = value;}
    }
    public string Qualification
    {
        get {return qualification;}
        set {qualification = value;}
    }
    public override void Work()
    {
        Console.WriteLine($"Разработчик {Name} выполняет работу над новым проектом");
    }
    public override void TakeVacation()
    {
        Console.WriteLine($"Разработчику {Name} со специализацией {Specialization} {Qualification} требуется согласовать отпуск с проектными сроками");
    }
    
}
public class Designer : Employee
{
    public Designer(string name, int age) : base(name, age)
    {
        DesignStyle = "абстракционизм";
    }
    private string designstyle;
    public string DesignStyle
    {
        get {return designstyle;}
        set {designstyle = value;}
    }
    public override void Work()
    {
        Console.WriteLine($"{Name} работает над дизайн проектами в стиле {DesignStyle}");
    }
    public void MakeNewDesign()
    {
        Console.WriteLine($"{Name} выполняет работу над новым дизайном");
    }
}
public class EmployeeCollection<W> where W : Employee
{
    private List<W> _employee = new List<W>();
    public void Add(W employee)
    {
        _employee.Add(employee);
    }
    public void Remove(W employee)
    {
        _employee.Remove(employee);
    }
    public void DisplayEmployee()
    {
        foreach (var employee in _employee)
        {
            employee.GetInfo();
            employee.Work();
        }
    }
}

Console.WriteLine("-------------------------------------------");
EmployeeCollection<Employee> employees = new EmployeeCollection<Employee>();
employees.Add(new Manager("Кирк", 42));
employees.Add(new Developer("Михаил", 34));
employees.Add(new Designer("Кирилл", 30));
employees.Add(new Employee("Юля", 26));

employees.DisplayEmployee();
```
