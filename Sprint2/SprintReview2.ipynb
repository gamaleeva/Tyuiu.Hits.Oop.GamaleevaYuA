```
public interface IReportable
{
    public void GenerateReport();
}
public interface ITeamLeader
{
    public void GetTeamMembers();
}

public class Employee
{
    public string Name {get; set;}
    public string Position{get; set;}
    public double Salary {get;set;}
    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник {Name} с должностью {Position} и зарплатой {Salary}");
    }
}
public class Manager : Employee
{
    public string Department {get;set;}
    public Manager(string name, string position, double salary, string department) : base(name, position, salary)
    {
        Department = department;
    }
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Менеджер {Name} управляет отделом {Department}");
    }
}
public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName {get;set;}
    public ProjectManager(string name, string position, double salary, string department, string projectname) : base(name, position, salary, department)
    {
        ProjectName = projectname;
    }
    public List<string> listofmembers = new List<string>();
    public void AddMember(string member)
    {
        listofmembers.Add(member);
    }
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Название проекта: {ProjectName}");
    }
    public void GenerateReport()
    {
        Console.WriteLine($"Работа над {ProjectName} успешно выполнена");
    }
    public void GetTeamMembers()
    {
        Console.WriteLine("Список членов команды:");
        foreach (string member in listofmembers)
        {
            Console.WriteLine(member);
        }
    }
}
Employee myEmployee = new Employee("Никита", "Программист", 64000.5);
Manager myManager = new Manager("Виктор", "Менеджер", 50000.4, "Отдел");
ProjectManager myProjManager = new ProjectManager("Павел", "Проектный менеджер", 60000, "отдел проектов", "Кассовые аппараты");
myProjManager.AddMember("Дарья");
myProjManager.AddMember("Михаил");
myEmployee.PrintInfo();
Console.WriteLine("------------------------------------------");
myManager.PrintInfo();
Console.WriteLine("------------------------------------------");
myProjManager.PrintInfo();
myProjManager.GenerateReport();
myProjManager.GetTeamMembers();

```
