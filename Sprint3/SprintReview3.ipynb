```
using System;
using System.Collections.Generic;

public interface IInstallable
{
    public void Install()
    {
        
    }
}
public abstract class Component : IInstallable
{
    public Component(string methodOfInstall, int price, string name)
    {
        
        MethodOfInstall = methodOfInstall;
        Price = price;
        Name = name;
    }
    public string MethodOfInstall {get;set;}
    public int Price {get; set;}
    public string Name{get; set;}
    public virtual void GetInfo() // Пекрытие
    {
        Console.WriteLine($"Компонент {Name} с ценой {Price} и методом установки {MethodOfInstall}");
    }
    public override string ToString() // Перегрузка
    {
        return $"Компонент {Name} с ценой {Price} и методом установки {MethodOfInstall}";
    }
}
public class CPU : Component, IInstallable
{
    public CPU(string methodOfInstall, int price, string name) : base(methodOfInstall, price, name)
    {
        
        MethodOfInstall = methodOfInstall;
        Price = price;
        Name = name;
    }
    public override void GetInfo()
    {
        Console.WriteLine($"Центральный процессор {Name} с ценой {Price} и методом установки {MethodOfInstall}");
    }
     public void Install()
    {
        Console.WriteLine($"Метод установки ЦП в компьютер {MethodOfInstall}");
    }
    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Price == b.Price && a.Name == b.Name && a.MethodOfInstall == b.MethodOfInstall;
    }
    public static bool operator !=(CPU a, CPU b) => !(a == b);
    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Price, Name, MethodOfInstall);
    }
    
}
public class RAM : Component, IInstallable
{
    public RAM(string methodOfInstall, int price, string name) : base(methodOfInstall, price, name)
    {
        
        MethodOfInstall = methodOfInstall;
        Price = price;
        Name = name;
    }
     public override void GetInfo()
    {
        Console.WriteLine($"Оперативная память {Name} с ценой {Price} и методом установки {MethodOfInstall}");
    }
     public void Install()
    {
        Console.WriteLine($"Метод установки оперативной памяти в компьютер {MethodOfInstall}");
    }
     public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Price == b.Price && a.Name == b.Name && a.MethodOfInstall == b.MethodOfInstall;
    }

    public static bool operator !=(RAM a, RAM b) => !(a == b);
    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Price, Name, MethodOfInstall);
    }
}
public class GPU : Component, IInstallable
{
    public GPU(string methodOfInstall, int price, string name) : base(methodOfInstall, price, name)
    {
        
        MethodOfInstall = methodOfInstall;
        Price = price;
        Name = name;
    }
     public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта {Name} с ценой {Price} и методом установки {MethodOfInstall}");
    }
     public void Install()
    {
        Console.WriteLine($"Метод установки видеокарты в компьютер {MethodOfInstall}");
    }
     public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Price == b.Price && a.Name == b.Name && a.MethodOfInstall == b.MethodOfInstall;
    }

    public static bool operator !=(GPU a, GPU b) => !(a == b);
    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Price, Name, MethodOfInstall);
    }
}
public class HDD : Component, IInstallable
{
    public HDD(string methodOfInstall, int price, string name) : base(methodOfInstall, price, name)
    {
        
        MethodOfInstall = methodOfInstall;
        Price = price;
        Name = name;
    }
     public override void GetInfo()
    {
        Console.WriteLine($"Жесткий диск {Name} с ценой {Price} и методом установки {MethodOfInstall}");
    }
     public void Install()
    {
        Console.WriteLine($"Метод установки жесткого диска в компьютер {MethodOfInstall}");
    }
     public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Price == b.Price && a.Name == b.Name && a.MethodOfInstall == b.MethodOfInstall;
    }

    public static bool operator !=(HDD a, HDD b) => !(a == b);
    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Price, Name, MethodOfInstall);
    }
    
}
public class Computer<T> where T : Component, IInstallable
{
    private List<T> components = new List<T>();
    public void AddComponent(T component)
    {
        components.Add(component);
    }
    public void RemoveComponent(T component)
    {
        components.Remove(component);
    }
    public void GetTotalPrice()
    {
        int totalPrice = 0;
        foreach (var component in components)
        {
            totalPrice += component.Price;
        }
        Console.WriteLine($"Стоимость всех компонентов коммпьютера {totalPrice}");
    }
}
public static class ComputerFactory<T> where T : Computer<Component>
{
    
    public static Computer<Component> AddOfficeComputer()
    {
        Computer<Component> computer = new Computer<Component>();
        computer.AddComponent(new CPU("-", 20000, "Центральный процессор"));
        computer.AddComponent(new RAM("-", 30000, "Оперативна память"));
        computer.AddComponent(new HDD("-", 35000, "Жесткий диск"));
        computer.AddComponent(new GPU("-", 20000, "Видеокарта"));
        return computer;
        
    }
     public static Computer<Component> AddGamingComputer()
    {
        Computer<Component> computer = new Computer<Component>();
        computer.AddComponent(new CPU("-", 20000, "Центральный процессор"));
        computer.AddComponent(new RAM("-", 30000, "Оперативная память"));
        computer.AddComponent(new HDD("-", 35000, "Жесткий диск"));
        computer.AddComponent(new GPU("-", 20000, "Видеокарта"));
        return computer;
    }
}
Computer<Component> pc = new Computer<Component>();
pc.AddComponent(new CPU("LGA1700", 45000, "Intel Core i7"));
pc.AddComponent(new RAM("DDR5",    12000, "Kingston 16GB"));
pc.AddComponent(new GPU("PCIe x16", 90000, "RTX 4070"));
pc.AddComponent(new HDD("SATA",     8000, "Seagate 1TB"));
pc.GetTotalPrice();
CPU cPU = new CPU("LGA1700", 45000, "Intel Core i7");
CPU cpU = new CPU("LGA1851", 35000, "Intel Core Ultra 5 245K");
cPU.GetInfo();
cPU.Install();
Console.WriteLine(cpU);
Console.WriteLine(cpU == cPU);
```
